```python
# sistema de caixa eletronico
def depositar(saldo):
  print(f'Seu saldo atual é {saldo}')
  valor_deposito = float(input("Digite a quantidade que deseja depositar: "))
  saldo = saldo + valor_deposito
  return saldo
  # fazer o código
  # informar ao usuario o saldo atual
  # perguntar qual valor quer depositar
  # fazer o algoritmo do deposito (saldo + deposito)
  # informar o novo saldo

def sacar(saldo):
  print(f"Seu saldo atua é de R$ {saldo:.2f}")
  valor_de_saque = float(input("Digite o valor a sacar: "))
  if valor_de_saque > saldo:
    print("Valor insuficiente na conta!")
  else:
    saldo = saldo - valor_de_saque
  return saldo
  # fazer o código
  # informar ao usuario o saldo atual
  # perguntar qual valor quer sacar
  # fazer o algoritmo do saque (saldo - saque)
  # informar o novo saldo
  #opcional - se o saldo for menor que o saque,
  # nao permitir
  # fazer o algoritmo do saque (saldo - saque)
  # informar o novo saldo
  #opcional - se o saldo for menor que o saque,
  # nao permitir
```

```python
saldo = float(input("Digite o valor do saldo: "))

print("===== BEM VINDO AO BANCO GEN ====\n")


while True:
  acao = int(input("Digite a operação: 'VER SALDO => 1' - 'SACAR => 2' - 'DEPOSITAR => 3'"))
  if acao == 1:
    print(f'Seu saldo atual é de R$ {saldo:.2f}')
  elif acao == 2:
    saldo = (sacar(saldo))
    print(f'Seu saldo atual é de R$ {saldo:.2f}')
  elif acao == 3:
    saldo = (depositar(saldo))
    print(f'Seu saldo atual é de R$ {saldo:.2f}')
  else:
    print("Ação inválida!")
    break

  reter = int(input("Deseja realizar uma operação novamente? 'SIM => 1' 'NÃO => 0'"))
  if reter == 0:
    print("Obrigado por utilizar nossos serviços!")
    break
```
